```python
import numpy as np  
import compress_planck
import new_desi_bao
##############################
data=np.loadtxt("data/pantheon_data_M.txt")
z_data_sn=data[:,0]

data_H = np.loadtxt("cc_data.txt")
z_dataH = data_H[:, 0]


data_pl = compress_planck.data_cmb # planck data. 


base_paths = ["lcdm_poly_cc+pla+bao"]

def compute_stats_from_chain(data, ndata):
        weights = data[:, 0]
        loglikes = -data[:, 1]*0.5
        samples = data[:, 2:]
        
        ndim = samples.shape[1]

        def aic(log_liklihood,ndim):

            res = -2 * log_liklihood + 2 * ndim
            return res

        def bic(log_liklihood,ndim,ndata):
            return -2 * log_liklihood + ndim * np.log(ndata)

        print("Ndim is",ndim)

        max_index = np.argmax(loglikes)
        max_loglike = loglikes[max_index]
        max_params = samples[max_index]

        chi2_val = -2 * max_loglike
        reduced_chi2 = chi2_val / (ndata - ndim)

        print("Max log likelihood: ", max_loglike)
        print("Max params: ", max_params)   
        print("Min chi_square: ", chi2_val)
        print("Reduced chi_square: ", reduced_chi2)
        print("AIC: ", aic(max_loglike, ndim))
        print("BIC: ", bic(max_loglike, ndim, ndata))



        return None



ndata = len(z_dataH) + len(new_desi_bao.z_desi_bao_eff) +  len(data_pl)

print(ndata)
data = np.loadtxt(f'chains/{base_paths[0]}.txt')
print("Data set is", base_paths[0])




stats = compute_stats_from_chain(data, ndata)
print(stats)
```
